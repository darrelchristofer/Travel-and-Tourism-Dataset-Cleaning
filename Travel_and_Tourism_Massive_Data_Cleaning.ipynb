### 2.3 Define values means "Missing"

```python
# Text values that means "Missing" seen from data uniqueness
NULL_TOKENS = {"", "null", "n/a", "na", "-", "?", "unknown", "#n/a", "none", "nan", "nil"}

WORD_NUMBERS = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5, "six": 6, "seven": 7,
                "eight": 8, "nine": 9, "ten": 10}

DATE_FORMATS = ["%Y-%m-%d", "%d/%m/%Y", "%d-%b-%Y", "%B %d, %Y", "%d.%m.%Y", "%Y/%m/%d", "%d %b %Y",
                "%d-%m-%y", "%Y-%m-%d %H:%M:%S", "%Y-%m-%dT%H:%M:%SZ"]
```

### 2.4 Define clean_text to Clean a Text Data

```python
# Strip spaces, Collapse spaces, and Turn it into real missing values

def clean_text(series):
    s = series.astype("string").str.strip().str.replace(r"\s+", " ", regex=True)
    return s.mask(s.str.lower().isin(NULL_TOKENS))
```

### 2.5 Define map_values for Labels and Clean them

```python
# Map Messy Labels to clean labels and print anything that could not be mapped

def map_values(series, mapping, column_name):
    s = clean_text(series).str.lower()
    mapped = s.map(mapping)
    unmapped = s[mapped.isna() & s.notna()].unique()
    if len(unmapped) > 0:
        print(f"Unmapped values in '{column_name}':", list(unmapped))
    return mapped
```

### 2.6 Define parse_dates, new format for Date

```python
# Try every date format found in the data. month_first=True reads 05/14/2024 as May 14

def parse_dates(series, month_first=False):
    s = clean_text(series)
    result = None
    for fmt in DATE_FORMATS:
        if month_first:
            fmt = fmt.replace("%d/%m/%Y", "%m/%d/%Y")
        parsed = pd.to_datetime(s, format=fmt, errors="coerce")
        result = parsed if result is None else result.fillna(parsed)
    return result
```

### 2.7 Define parse_money, numbers format for Money and Currency

```python
# Turn text like 'Rp 1.250.000,-', '$1,250.50', '250,50' or '¥42,000' into a number

def parse_money(value, zero_decimal = False):
    if pd.isna(value):
        return np.nan
    text = str(value).strip().replace(",-", "")
    negative = "-" in text
    number = re.sub(r"[^0-9.,]", "", text)
    if not re.search(r"\d", number):
        return np.nan
    if "." in number and "," in number:
        if number.rfind(".") > number.rfind(","):
            number = number.replace(",", "")                     # 1,250.50
        else:
            number = number.replace(".", "").replace(",", ".")   # 1.250,50
    elif number.count(".") > 1:
        number = number.replace(".", "")                         # 1.250.000
    elif number.count(",") > 1:
        number = number.replace(",", "")                         # 1,250,000
    elif "," in number:
        decimals = len(number.split(",")[1])
        number = number.replace(",", ".") if decimals in (1, 2) else number.replace(",", "")
    elif "." in number:
        decimals = len(number.split(".")[1])
        if decimals == 3 and zero_decimal:
            number = number.replace(".", "")                     # Rp 750.000
    amount = float(number)
    return -amount if negative else amount
```

### 2.8 Define parse number, for Numbers hidden in Text

```python
# Find Numbers in hidden Text (Turn '3 nights', '3N', 'three', '3.0' into 3)

def parse_number(value):
    if pd.isna(value):
        return np.nan
    text = str(value).strip().lower()
    if text in WORD_NUMBERS:
        return WORD_NUMBERS[text]
    match = re.search(r"-?\d+(\.\d+)?", text)
    return float(match.group()) if match else np.nan
```

### 2.9 Define standardize_customer_id and standardize_property_id 

```python
# Standardize Customer ID (c000123, C-000123, 123, C123 -> C000123)

def standardize_customer_id(series):
    ids = clean_text(series).str.upper().str.replace(r"[^A-Z0-9]", "", regex=True)
    numbers = ids.str.extract(r"^C?(\d+)$")[0]
    return "C" + numbers.str.zfill(6)

# Standardize Property ID (pr-0012, PR0012, PR12, PR_0012 -> PR-0012)

def standardize_property_id(series):
    ids = clean_text(series).str.upper().str.replace(r"[^A-Z0-9]", "", regex=True)
    numbers = ids.str.extract(r"^PR(\d+)$")[0]
    return "PR-" + numbers.str.zfill(4)
```

### 2.10 Define fix_case for Mixed-Cased values

```python
# Converts values to Title Case

def fix_case(series):
    s = series.astype("string")
    wrong_case = ((s == s.str.upper()) | (s == s.str.lower())).fillna(False)
    return s.where(~wrong_case, s.str.title())
```

#### 3.3.2 Clean full_name

#### 3.3.3 Clean email
Lowercase the email, keep only the first email, fix typo domains, repair a missing "@", and remove placeholder and invalid emails.

```python
DOMAIN_FIXES = {"gmial.com": "gmail.com", "gmail.con": "gmail.com", "gmail.co": "gmail.com",
                "yahooo.com": "yahoo.com", "hotmial.com": "hotmail.com", "outlok.com": "outlook.com"}
KNOWN_DOMAINS = ["yahoo.co.id", "gmail.com", "yahoo.com", "outlook.com", "hotmail.com", "icloud.com", "proton.me"]
PLACEHOLDER_EMAILS = {"none@none.com", "noemail@email.com", "-@-.com", "test@test.com"}

import re

def fix_email(value):
    if pd.isna(value):
        return pd.NA
    email = str(value).strip().lower().split(";")[0].strip()
    if email in PLACEHOLDER_EMAILS:
        return pd.NA
    if "@" not in email:
        for domain in KNOWN_DOMAINS:
            if email.endswith(domain):
                email = email[: -len(domain)] + "@" + domain
                break
    if "@" in email:
        local, domain = email.rsplit("@", 1)
        email = local + "@" + DOMAIN_FIXES.get(domain, domain)
    if not re.match(r"^[\w.+-]+@[\w-]+(\.[\w-]+)+$", email):
        return pd.NA
    return email


before = customers["email"].astype("string")
customers["email"] = customers["email"].apply(fix_email).astype("string")

log_step("customers", "email", "standardized/repaired emails", (before.str.strip() != customers["email"]).fillna(False).sum())
log_step("customers", "email", "set placeholder or invalid emails to missing", (before.notna() & customers["email"].isna()).sum())
```

#### 3.3.5 Standardize phone numbers
Convert phone numbers to international format (+62812...).

- `NUMBER_LENGTHS` is the length of the national number (without country code or leading 0), as seen in the profiling.
- Junk values like "12345", "0" or "call after 5pm" become missing.
- Local format `0812...` → `+62812...`
- `62812...` or `(62) 812...` → `+62812...`
- `812...` without a prefix → `+62812...`

```python
CALLING_CODES = {"Indonesia": "62", "Singapore": "65", "Malaysia": "60", "Australia": "61", "United Kingdom": "44",
                 "United States": "1", "Japan": "81", "South Korea": "82", "Germany": "49", "France": "33", "India": "91"}

NUMBER_LENGTHS = {"Indonesia": (10, 11)}


def standardize_phone(raw, nationality):
    if pd.isna(raw):
        return pd.NA
    text = str(raw).strip()
    digits = re.sub(r"\D", "", text)
    if len(digits) < 8:
        return pd.NA
    if text.startswith("+"):
        return "+" + digits
    code = CALLING_CODES.get(nationality)
    if code is None:
        return pd.NA
    lengths = NUMBER_LENGTHS.get(nationality, (9,))
    if digits.startswith("0") and len(digits) - 1 in lengths:
        return "+" + code + digits[1:]
    if digits.startswith(code) and len(digits) - len(code) in lengths:
        return "+" + digits
    if len(digits) in lengths:
        return "+" + code + digits
    return pd.NA


before = customers["phone"].astype("string")
customers["phone"] = [standardize_phone(p, n) for p, n in zip(customers["phone"], customers["nationality"])]
customers["phone"] = customers["phone"].astype("string")

log_step("customers", "phone", "converted phones to +<country code><number>", customers["phone"].notna().sum())
log_step("customers", "phone", "set junk phone values to missing", (before.notna() & customers["phone"].isna()).sum())
print(customers["phone"].head(10).tolist())
```

#### 3.3.6 Parse date_of_birth
Parse `date_of_birth` and remove impossible values (future dates, dummy 1900/1970 dates, ages over 100 or under 18).

Two-digit years like "05-03-62" are read as 2062, so they are moved back 100 years.

```python
dob_text = clean_text(customers["date_of_birth"])
dob = parse_dates(customers["date_of_birth"])

two_digit_year = dob_text.str.match(r"^\d{2}-\d{2}-\d{2}$").fillna(False)
future_two_digit = two_digit_year & (dob > EXTRACT_DATE).fillna(False)
dob.loc[future_two_digit] = dob[future_two_digit] - pd.DateOffset(years=100)
log_step("customers", "date_of_birth", "fixed two-digit years read in the wrong century", future_two_digit.sum())

dummy_dates = dob.isin([pd.Timestamp("1900-01-01"), pd.Timestamp("1970-01-01")])
age = (EXTRACT_DATE - dob).dt.days / 365.25
impossible_age = ~age.between(18, 100) & dob.notna()

log_step("customers", "date_of_birth", "set dummy dates (1900-01-01 / 1970-01-01) to missing", dummy_dates.sum())
log_step("customers", "date_of_birth", "set impossible ages (future, <18, >100) to missing", (impossible_age & ~dummy_dates).sum())
log_step("customers", "date_of_birth", "unparseable dates set to missing", (dob_text.notna() & dob.isna()).sum())

customers["date_of_birth"] = dob.mask(dummy_dates | impossible_age).dt.normalize()
```

#### 3.3.10 Parse annual_income_usd
Convert income text ("$52,000", "52k", "USD 52000", "Rp 832.000.000") into numbers. Incomes above 1,000,000 (e.g. 27,000,000) have extra zeros and are not real incomes, so they are set to missing.

#### 3.3.12 Find duplicate customers (entity resolution)
Find the same person registered under different customer IDs.

- Rule: two records are the same person when at least 2 of these 3 match: email, phone, date of birth.
- Emails alone are not enough, because different people can share a common email like budi.santoso@gmail.com.
- Every duplicate ID is linked to one master ID (the smallest / oldest ID).

```python
from itertools import combinations

pair_matches = {}
for key in ["email", "phone"]:
    groups = customers.dropna(subset=[key]).groupby(key)["customer_id"].apply(list)
    for ids in groups[groups.str.len() > 1]:
        for a, b in combinations(sorted(ids), 2):
            pair_matches.setdefault((a, b), set()).add(key)

dob_lookup = customers.set_index("customer_id")["date_of_birth"]
same_person_pairs = []
for (a, b), keys in pair_matches.items():
    votes = len(keys)
    if pd.notna(dob_lookup[a]) and dob_lookup[a] == dob_lookup[b]:
        votes += 1
    if votes >= 2:
        same_person_pairs.append((a, b))

master_of = {}

def find_master(customer_id):
    while master_of.get(customer_id, customer_id) != customer_id:
        customer_id = master_of[customer_id]
    return customer_id

for a, b in same_person_pairs:
    master_a, master_b = find_master(a), find_master(b)
    if master_a != master_b:
        master_of[max(master_a, master_b)] = min(master_a, master_b)

customer_id_map = {cid: find_master(cid) for cid in customers["customer_id"]}
customers["master_customer_id"] = customers["customer_id"].map(customer_id_map)

log_step("customers", "customer_id", "same person found under a second customer ID (merged)", (customers["customer_id"] != customers["master_customer_id"]).sum())
print("Example matches:", same_person_pairs[:5])
```

#### 3.3.13 Merge duplicate customers
Merge each duplicate person into one row: the master row comes first, and its missing fields are filled from the duplicate.

```python
customers["is_master"] = customers["customer_id"] == customers["master_customer_id"]
customers = customers.sort_values(["master_customer_id", "is_master"], ascending=[True, False])

first_signup = customers.groupby("master_customer_id")["signup_date"].min()
merged_ids = customers.groupby("master_customer_id")["customer_id"].apply(lambda x: ";".join(x[1:]))

customers_clean = customers.groupby("master_customer_id", as_index=False).first()
customers_clean["customer_id"] = customers_clean["master_customer_id"]
customers_clean["signup_date"] = customers_clean["master_customer_id"].map(first_signup)
customers_clean["merged_customer_ids"] = customers_clean["master_customer_id"].map(merged_ids).replace("", pd.NA)
customers_clean = customers_clean.drop(columns=["master_customer_id", "is_master"])

print("Customers before merge:", len(customers), "| after merge:", len(customers_clean))
```

#### 4.2.9 Parse and fix coordinates
Parse coordinates (decimal commas, "8.5069° S") and fix swapped, flipped, 0,0 and missing values.

The reference point of each city is the median of all its properties.

```python
def parse_coordinate(value):
    if pd.isna(value):
        return np.nan
    text = str(value).strip().upper().replace(",", ".")
    match = re.search(r"-?\d+(\.\d+)?", text)
    if not match:
        return np.nan
    number = float(match.group())
    if text.endswith("S") or text.endswith("W"):
        number = -abs(number)
    return number


lat = properties["latitude"].apply(parse_coordinate)
lon = properties["longitude"].apply(parse_coordinate)

null_island = (lat == 0) & (lon == 0)
lat, lon = lat.mask(null_island), lon.mask(null_island)
log_step("properties", "latitude/longitude", "set 0,0 coordinates to missing", null_island.sum())

city_lat = lat.groupby(properties["city"]).transform("median")
city_lon = lon.groupby(properties["city"]).transform("median")

def is_near(a, b):
    return ((a - city_lat).abs() < 1) & ((b - city_lon).abs() < 1)

far_away = ~is_near(lat, lon) & lat.notna()
swapped = far_away & is_near(lon, lat)
flipped = far_away & ~swapped & is_near(-lat, lon)
still_wrong = far_away & ~swapped & ~flipped

new_lat = lat.mask(swapped, lon).mask(flipped, -lat).mask(still_wrong)
new_lon = lon.mask(swapped, lat).mask(still_wrong)

log_step("properties", "latitude/longitude", "fixed swapped latitude/longitude", swapped.sum())
log_step("properties", "latitude/longitude", "fixed flipped latitude sign", flipped.sum())
log_step("properties", "latitude/longitude", "set coordinates far from their city to missing", still_wrong.sum())

properties["latitude"] = new_lat.round(6)
properties["longitude"] = new_lon.round(6)
```

#### 4.2.10 Clean check_in_time, total_rooms and guest_score
- `check_in_time`: "2 PM", "14.00", "1400" → "14:00"
- `total_rooms`: remove negative or impossible room counts
- `guest_score`: "84%" → 8.4, and "84" → 8.4 (scale error)

```python
def parse_time(value):
    if pd.isna(value):
        return pd.NA
    text = str(value).strip().lower().replace(" ", "")
    match = re.match(r"^(\d{1,2})[:.]?(\d{2})?(?::\d{2})?(am|pm)?$", text)
    if not match:
        return pd.NA
    hour = int(match.group(1))
    minute = int(match.group(2) or 0)
    if match.group(3) == "pm" and hour < 12:
        hour += 12
    return f"{hour:02d}:{minute:02d}"


properties["check_in_time"] = properties["check_in_time"].apply(parse_time).astype("string")

rooms = properties["total_rooms"].apply(parse_number)
invalid_rooms = (rooms <= 0) | (rooms > 1000)
log_step("properties", "total_rooms", "set negative or impossible room counts to missing", invalid_rooms.sum())
properties["total_rooms"] = rooms.mask(invalid_rooms).astype("Int64")


def parse_score(value):
    if pd.isna(value):
        return np.nan
    text = str(value).strip().replace(",", ".")
    number = parse_number(text)
    if pd.isna(number):
        return np.nan
    if "%" in text:
        return number / 10
    if 10 < number <= 100 and number == int(number):
        return number / 10
    return number


score = properties["guest_score"].apply(parse_score)
invalid_score = (score <= 0) | (score > 10)
log_step("properties", "guest_score", "set zero or out-of-range scores to missing", invalid_score.sum())
properties["guest_score"] = score.mask(invalid_score).round(1)

print(properties[["check_in_time", "total_rooms", "guest_score"]].describe(include="all"))
```

#### 5.3.5 Evidence: date order by channel
Which date order does each channel use? Look at slash dates like 05/14/2024.

If the FIRST number is above 12 it must be the day (day-first). If the SECOND number is above 12 it is month-first.

```python
slash_parts = clean_text(bookings["check_in"]).str.extract(r"^(\d{2})/(\d{2})/\d{4}$").astype(float)
evidence = pd.DataFrame({"channel": bookings["channel"],
                         "first_number_over_12": slash_parts[0] > 12,
                         "second_number_over_12": slash_parts[1] > 12})
print(evidence.groupby("channel", dropna=False)[["first_number_over_12", "second_number_over_12"]].sum())
print("\nConclusion: Booking.com writes dates month-first (MM/DD/YYYY), every other channel writes day-first (DD/MM/YYYY).")
```

#### 5.3.6 Parse nights and guests
Convert nights and guests text ("3 nights", "three", "2A+1C", "2 adults") into numbers.

```python
nights_raw = bookings["nights"].apply(parse_number)


def parse_guests(value):
    if pd.isna(value):
        return np.nan
    text = str(value).strip().lower()
    match = re.match(r"^(\d+)a\+(\d+)c$", text)
    if match:
        return int(match.group(1)) + int(match.group(2))
    return parse_number(text)


guests = bookings["guests"].apply(parse_guests)
invalid_guests = (guests < 1) | (guests > 12)
log_step("bookings", "guests", "set impossible guest counts (0, negative, 40, 99) to missing", invalid_guests.sum())
bookings["guests"] = guests.mask(invalid_guests).astype("Int64")
```

#### 5.3.7 Parse check_in and check_out
Parse `check_in` and `check_out` using the date order of each channel.

- For ambiguous dates (e.g. 03/07/2024) on Booking.com / unknown-channel rows, or stays longer than 30 nights, try both readings and pick the one where the gap between the two dates matches the nights column.
- If nothing matches, pick the first reading that gives a realistic stay of 1-30 nights.
- The gap is compared without its sign, because swapped check_in/check_out are fixed afterwards.
- Remove impossible years (typos like 2203 or 1923).
- Swap dates when check_out is earlier than check_in.
- Fill one missing date from the other date + nights.

```python
is_booking_com = (bookings["channel"] == "Booking.com").fillna(False)
no_channel = bookings["channel"].isna()


def parse_by_channel(series):
    day_first = parse_dates(series)
    month_first = parse_dates(series, month_first=True)
    preferred = month_first.where(is_booking_com, day_first)
    fallback = day_first.where(is_booking_com, month_first)
    return preferred.fillna(fallback), day_first, month_first


check_in, ci_day_first, ci_month_first = parse_by_channel(bookings["check_in"])
check_out, co_day_first, co_month_first = parse_by_channel(bookings["check_out"])

ambiguous = (((ci_day_first != ci_month_first) & ci_day_first.notna() & ci_month_first.notna())
             | ((co_day_first != co_month_first) & co_day_first.notna() & co_month_first.notna())).fillna(False)
stay = (check_out - check_in).dt.days
needs_check = ambiguous & (is_booking_com | no_channel | ~stay.abs().between(1, 30))
resolved = 0
for i in bookings.index[needs_check]:
    ci_options = [ci_month_first[i], ci_day_first[i]] if is_booking_com[i] else [ci_day_first[i], ci_month_first[i]]
    co_options = [co_month_first[i], co_day_first[i]] if is_booking_com[i] else [co_day_first[i], co_month_first[i]]
    options = [(ci, co) for ci in ci_options for co in co_options if pd.notna(ci) and pd.notna(co)]
    matches_nights = [o for o in options if abs((o[1] - o[0]).days) == nights_raw[i]]
    realistic = [o for o in options if 1 <= abs((o[1] - o[0]).days) <= 30]
    if matches_nights:
        choice = matches_nights[0]
    elif realistic and not (1 <= abs(stay[i]) <= 30):
        choice = realistic[0]
    else:
        continue
    if choice[0] != check_in[i] or choice[1] != check_out[i]:
        check_in.loc[i], check_out.loc[i] = choice
        resolved += 1
log_step("bookings", "check_in/check_out", "resolved ambiguous day/month order using nights", resolved)

valid_range = (pd.Timestamp("2022-01-01"), pd.Timestamp("2026-12-31"))
bad_ci = ~check_in.between(*valid_range) & check_in.notna()
bad_co = ~check_out.between(*valid_range) & check_out.notna()
log_step("bookings", "check_in/check_out", "set impossible years to missing", bad_ci.sum() + bad_co.sum())
check_in, check_out = check_in.mask(bad_ci), check_out.mask(bad_co)
log_step("bookings", "check_in/check_out", "unparseable dates (e.g. 2023-02-30, TBD) left missing",
         (clean_text(bookings["check_in"]).notna() & check_in.isna()).sum() + (clean_text(bookings["check_out"]).notna() & check_out.isna()).sum())

swap = (check_out < check_in).fillna(False)
check_in, check_out = check_in.mask(swap, check_out), check_out.mask(swap, check_in)
log_step("bookings", "check_in/check_out", "swapped check_in and check_out (check_out was earlier)", swap.sum())

valid_nights = nights_raw.where(nights_raw.between(1, 60))
fill_ci = check_in.isna() & check_out.notna() & valid_nights.notna()
fill_co = check_out.isna() & check_in.notna() & valid_nights.notna()
check_in = check_in.mask(fill_ci, check_out - pd.to_timedelta(valid_nights, unit="D"))
check_out = check_out.mask(fill_co, check_in + pd.to_timedelta(valid_nights, unit="D"))
log_step("bookings", "check_in/check_out", "filled one missing date using the other date and nights", fill_ci.sum() + fill_co.sum())

bookings["check_in"] = check_in
bookings["check_out"] = check_out
```

#### 5.3.8 Recalculate nights
Recalculate nights from the dates (dates are the source of truth), and use the nights column only when dates are missing.

```python
nights_from_dates = (bookings["check_out"] - bookings["check_in"]).dt.days
mismatch = (nights_from_dates != nights_raw) & nights_from_dates.notna() & nights_raw.notna()
log_step("bookings", "nights", "nights column did not match dates (replaced by date difference)", mismatch.sum())

bookings["nights"] = nights_from_dates.fillna(valid_nights).astype("Int64")
print(bookings["nights"].describe())
```

#### 5.3.9 Parse booked_at
Parse `booked_at`: local time (WIB, UTC+7), UTC timestamps ending in "Z", or date only.

For ambiguous date-only values on Booking.com, choose the reading that is not after check-in.

```python
booked_text = clean_text(bookings["booked_at"])
is_utc = booked_text.str.endswith("Z").fillna(False)
booked_utc = pd.to_datetime(booked_text.where(is_utc), format="%Y-%m-%dT%H:%M:%SZ", errors="coerce") + pd.Timedelta(hours=7)
log_step("bookings", "booked_at", "converted UTC timestamps to local time (WIB, +7 hours)", is_utc.sum())

booked_local, booked_day_first, booked_month_first = parse_by_channel(bookings["booked_at"])
booked_at = booked_utc.fillna(booked_local)

other_reading = booked_day_first.where(is_booking_com, booked_month_first)
use_other = (booked_at.dt.normalize() > bookings["check_in"]) & (other_reading <= bookings["check_in"]) & ~is_utc
booked_at = booked_at.mask(use_other.fillna(False), other_reading)

booked_after_checkin = (booked_at.dt.normalize() > bookings["check_in"]).fillna(False)
log_step("bookings", "booked_at", "set booking dates after check-in to missing", booked_after_checkin.sum())
bookings["booked_at"] = booked_at.mask(booked_after_checkin)
```

#### 5.3.10 Parse currency and total_amount
Currency priority: symbol inside the amount text (Rp, $, ¥...) > currency column > property currency.

```python
def currency_from_text(value):
    if pd.isna(value):
        return pd.NA
    text = str(value)
    checks = [("IDR", r"Rp|IDR"), ("SGD", r"S\$|SGD"), ("AUD", r"A\$|AUD"), ("USD", r"US\$|USD"), ("MYR", r"RM|MYR"),
              ("AED", r"AED"), ("EUR", r"€|EUR"), ("JPY", r"¥|JPY"), ("KRW", r"₩|KRW"), ("VND", r"₫|VND"),
              ("THB", r"฿|THB"), ("USD", r"\$")]
    for currency, pattern in checks:
        if re.search(pattern, text):
            return currency
    return pd.NA


label = clean_text(bookings["currency"]).str.upper().replace({"RP": "IDR", "RUPIAH": "IDR", "US$": "USD", "$": "USD"})
symbol = bookings["total_amount"].apply(currency_from_text).astype("string")
property_currency = bookings["property_id"].map(properties_clean.set_index("property_id")["rate_currency"])

disagree = symbol.notna() & label.notna() & (symbol != label)
log_step("bookings", "currency", "currency column contradicted the symbol in the amount (symbol used)", disagree.fillna(False).sum())
log_step("bookings", "currency", "missing currency filled from amount symbol or property", label.isna().sum())

bookings["currency"] = symbol.fillna(label).fillna(property_currency)
bookings["total_amount"] = [parse_money(v, c in ZERO_DECIMAL) for v, c in zip(bookings["total_amount"], bookings["currency"])]
print(bookings["currency"].value_counts(dropna=False))
```

#### 5.3.11 Fix currency mislabels and amount errors
Compare each amount with the property's typical spend per night (learned from all bookings of the same property, so it also works when the catalog price is missing).

1. Negative amounts are sign errors, zero amounts are missing.
2. Typical spend per night (USD) for each property = median over its bookings (the median ignores the few bad values).
3. Amounts tagged USD that are really local currency (ratio around 16,000 for IDR) are relabelled.
4. Amounts tagged with a local currency that are really USD (e.g. 43.70 "IDR" or 289.65 "THB" for a hotel room) are relabelled.
5. Amounts with extra zeros (x100) are divided by 100.

```python
negative = bookings["total_amount"] < 0
zero = bookings["total_amount"] == 0
log_step("bookings", "total_amount", "fixed negative amounts (sign error)", negative.sum())
log_step("bookings", "total_amount", "set zero amounts to missing", zero.sum())
bookings["total_amount"] = bookings["total_amount"].abs().mask(zero)

nights_float = bookings["nights"].astype(float)
spend_per_night = bookings["total_amount"] / bookings["currency"].map(RATES) / nights_float
typical_per_night = spend_per_night.groupby(bookings["property_id"]).transform("median")
expected_usd = nights_float * typical_per_night

ratio = bookings["total_amount"] / bookings["currency"].map(RATES) / expected_usd
mislabel = ((bookings["currency"] == "USD") & (property_currency != "USD") & (ratio > 1000)).fillna(False)
bookings.loc[mislabel, "currency"] = property_currency[mislabel]
log_step("bookings", "currency", "amount tagged USD but clearly in local currency (relabelled)", mislabel.sum())

ratio = bookings["total_amount"] / bookings["currency"].map(RATES) / expected_usd
really_usd = ((bookings["currency"] != "USD") & (ratio < 0.2) & ((ratio * bookings["currency"].map(RATES)).between(0.3, 3))).fillna(False)
bookings.loc[really_usd, "currency"] = "USD"
log_step("bookings", "currency", "amount tagged local currency but clearly USD (relabelled)", really_usd.sum())

ratio = bookings["total_amount"] / bookings["currency"].map(RATES) / expected_usd
too_high = (ratio > 20).fillna(False)
fixable = too_high & (ratio / 100).between(0.1, 5)
bookings.loc[fixable, "total_amount"] = bookings.loc[fixable, "total_amount"] / 100
bookings.loc[too_high & ~fixable, "total_amount"] = np.nan
log_step("bookings", "total_amount", "divided extra-zero outliers by 100", fixable.sum())
log_step("bookings", "total_amount", "set unfixable outliers to missing", (too_high & ~fixable).sum())

bookings["amount_usd"] = (bookings["total_amount"] / bookings["currency"].map(RATES)).round(2)
print(bookings["amount_usd"].describe())
```

#### 5.3.12 Impute missing amounts
Missing amounts = nights x the typical spend per night for the same property and room type (fallback: same property, any room type). Learned from the cleaned amounts, not from the catalog price.

```python
spend_per_night = bookings["amount_usd"] / nights_float
typical_property_room = spend_per_night.groupby([bookings["property_id"], bookings["room_type"]]).transform("median")
typical_property = spend_per_night.groupby(bookings["property_id"]).transform("median")
estimate = (nights_float * typical_property_room.fillna(typical_property)).round(2)

bookings["amount_imputed"] = bookings["amount_usd"].isna() & estimate.notna()
bookings["amount_usd"] = bookings["amount_usd"].fillna(estimate)
log_step("bookings", "amount_usd", "imputed missing amounts from the same property and room type", bookings["amount_imputed"].sum())
print("Bookings still without an amount:", bookings["amount_usd"].isna().sum())
```

#### 5.3.13 Standardize promo_code and discount_pct
Convert discounts ("10%", "0.1", "10 %") into one number (10). Each promo code always gives the same discount, so it is learned from the data.

```python
promo = clean_text(bookings["promo_code"]).str.upper().str.replace(r"[\s-]", "", regex=True)
bookings["promo_code"] = promo.fillna("NONE")


def parse_discount(value):
    number = parse_number(str(value).replace("%", "")) if pd.notna(value) else np.nan
    if pd.notna(number) and 0 < number <= 1:
        number = number * 100
    return number


discount = bookings["discount_pct"].apply(parse_discount)

code_discount = discount[discount > 0].groupby(bookings["promo_code"]).agg(lambda x: x.mode().iloc[0])
code_discount["NONE"] = 0
print("Discount per promo code:")
print(code_discount)

expected_discount = bookings["promo_code"].map(code_discount)
filled = expected_discount.gt(0) & discount.isna()
conflict = expected_discount.notna() & discount.notna() & (discount.round(0) != expected_discount)
log_step("bookings", "discount_pct", "filled missing discount from promo code", filled.sum())
log_step("bookings", "discount_pct", "fixed discount that conflicted with promo code", conflict.sum())
bookings["discount_pct"] = expected_discount.fillna(discount).astype("Int64")
```

#### 5.3.14 Parse cancel_date, infer missing status and fix logical conflicts
Missing status is inferred: cancel date → Cancelled, future check-in → Confirmed, has rating → Completed.

```python
cancel_date, cd_day_first, cd_month_first = parse_by_channel(bookings["cancel_date"])
other_reading = cd_day_first.where(is_booking_com, cd_month_first)
outside = (cancel_date > bookings["check_in"]) | (cancel_date < bookings["booked_at"].dt.normalize())
inside_other = (other_reading <= bookings["check_in"]) & (other_reading >= bookings["booked_at"].dt.normalize())
cancel_date = cancel_date.mask((outside & inside_other).fillna(False), other_reading)

missing_status = bookings["status"].isna()
inferred = np.select([cancel_date.notna(), bookings["check_in"] > EXTRACT_DATE, clean_text(bookings["guest_rating"]).notna()],
                     ["Cancelled", "Confirmed", "Completed"], default="Unknown")
bookings.loc[missing_status, "status"] = pd.Series(inferred, index=bookings.index)[missing_status]
log_step("bookings", "status", "inferred missing status from other columns", missing_status.sum())

not_cancelled = bookings["status"] != "Cancelled"
cancel_on_active = not_cancelled & cancel_date.notna()
cancel_after_checkin = (cancel_date > bookings["check_in"]).fillna(False)
log_step("bookings", "cancel_date", "removed cancel date on non-cancelled bookings", cancel_on_active.sum())
log_step("bookings", "cancel_date", "set cancel dates after check-in to missing", (cancel_after_checkin & ~cancel_on_active).sum())
bookings["cancel_date"] = cancel_date.mask(cancel_on_active | cancel_after_checkin)
print("Cancelled bookings without a cancel date:", ((bookings["status"] == "Cancelled") & bookings["cancel_date"].isna()).sum())
```

#### 5.3.15 Clean guest_rating
Convert ratings ("4/5", "4.0", "four", "8" on a 10-point scale) into a 1-5 number and remove ratings on stays that never happened.

```python
rating = bookings["guest_rating"].apply(parse_number)
ten_point = rating.between(6, 10)
log_step("bookings", "guest_rating", "converted 10-point ratings to 5-point (divided by 2)", ten_point.sum())
rating = rating.mask(ten_point, rating / 2)
zero_rating = rating == 0
log_step("bookings", "guest_rating", "set zero ratings to missing", zero_rating.sum())
rating = rating.mask(zero_rating)

not_completed = bookings["status"] != "Completed"
log_step("bookings", "guest_rating", "removed ratings on bookings that were not completed", (not_completed & rating.notna()).sum())
bookings["guest_rating"] = rating.mask(not_completed).round(0).astype("Int64")
```

#### 5.3.16 Clean free text: review and special_requests
Remove HTML, `<br>`, line breaks, extra spaces and placeholders.

```python
def clean_free_text(series):
    s = series.astype("string").apply(lambda x: html.unescape(x) if pd.notna(x) else x).astype("string")
    s = s.str.replace(r"<br\s*/?>", " ", regex=True).str.replace(r"[\r\n]+", " ", regex=True)
    return clean_text(s)


review = clean_free_text(bookings["review"])
review_on_not_completed = not_completed & review.notna()
log_step("bookings", "review", "removed reviews on bookings that were not completed", review_on_not_completed.sum())
bookings["review"] = review.mask(review_on_not_completed)

requests = clean_free_text(bookings["special_requests"])
requests = requests.mask(requests.str.lower().isin(["no", "nil"]))
all_upper = (requests == requests.str.upper()).fillna(False)
bookings["special_requests"] = requests.where(~all_upper, requests.str.capitalize())

print(bookings[["review", "special_requests"]].dropna().head())
```

#### 5.3.17 Remove near-duplicate bookings
Same customer, property, dates, room, guests and amount, but a new booking_id (keep the earliest one).

```python
near_dup_columns = ["customer_id", "property_id", "check_in", "check_out", "room_type", "guests", "total_amount", "channel", "status"]
bookings = bookings.sort_values("booked_at")
near_duplicates = bookings.duplicated(subset=near_dup_columns, keep="first")
log_step("bookings", "(row)", "removed near-duplicate bookings (re-submitted with a new ID)", near_duplicates.sum())
bookings = bookings[~near_duplicates].sort_values("booking_id").reset_index(drop=True)
print("Bookings shape:", bookings.shape)
```

#### 5.3.18 Add derived columns and finalize bookings
Add useful derived columns (lead time, check-in month, cancelled flag) and select the final columns.

```python
bookings["lead_time_days"] = (bookings["check_in"] - bookings["booked_at"].dt.normalize()).dt.days.astype("Int64")
bookings["check_in_month"] = bookings["check_in"].dt.to_period("M").astype("string")
bookings["is_cancelled"] = bookings["status"] == "Cancelled"
bookings["channel"] = bookings["channel"].fillna("Unknown")

bookings_clean = bookings[["booking_id", "customer_id", "property_id", "booked_at", "check_in", "check_out", "nights", "guests",
                           "room_type", "channel", "status", "is_cancelled", "total_amount", "currency", "amount_usd",
                           "amount_imputed", "payment_method", "payment_status", "promo_code", "discount_pct",
                           "travel_purpose", "special_requests", "cancel_date", "guest_rating", "review",
                           "lead_time_days", "check_in_month", "customer_known", "property_known"]]
print("Bookings clean shape:", bookings_clean.shape)
print(bookings_clean.isna().sum())
bookings_clean.head()
```

## 6. Validation checks

```python
# Run validation rules on the three clean tables

def check(description, passed):
    print(f"{'PASS' if passed else 'FAIL'} - {description}")


b, c, p = bookings_clean, customers_clean, properties_clean

check("booking_id is unique", b["booking_id"].is_unique)
check("check_out is after check_in", ((b["check_out"] > b["check_in"]) | b["check_in"].isna() | b["check_out"].isna()).all())
check("nights between 1 and 30", b["nights"].dropna().between(1, 30).all())
check("guests between 1 and 12", b["guests"].dropna().between(1, 12).all())
check("amount_usd is positive", (b["amount_usd"].dropna() > 0).all())
check("guest_rating between 1 and 5", b["guest_rating"].dropna().between(1, 5).all())
check("ratings only on completed stays", b.loc[b["guest_rating"].notna(), "status"].eq("Completed").all())
check("cancel_date only on cancelled bookings", b.loc[b["cancel_date"].notna(), "status"].eq("Cancelled").all())
check("booked_at is not after check_in", ((b["booked_at"].dt.normalize() <= b["check_in"]) | b["booked_at"].isna() | b["check_in"].isna()).all())
check("customer_id is unique", c["customer_id"].is_unique)
check("customer emails are valid", c["email"].dropna().str.match(r"^[\w.+-]+@[\w-]+(\.[\w-]+)+$").all())
check("customer ages between 18 and 100", ((EXTRACT_DATE - c["date_of_birth"]).dt.days.dropna() / 365.25).between(18, 100).all())
check("property_id is unique", p["property_id"].is_unique)
check("star_rating between 1 and 5", p["star_rating"].dropna().between(1, 5).all())
check("guest_score between 0 and 10", p["guest_score"].dropna().between(0, 10).all())
check("latitude between -90 and 90", p["latitude"].dropna().between(-90, 90).all())

print("-" * 100)
print(f"Bookings linked to a known customer: {b['customer_known'].mean():.1%}")
print(f"Bookings linked to a known property: {b['property_known'].mean():.1%}")
```
